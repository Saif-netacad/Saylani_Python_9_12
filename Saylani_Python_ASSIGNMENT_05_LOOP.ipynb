### 9.

with 3 and having a common difference of 4.
The program should output the following sequence:
3 7 11 15 19 23 27 31

```python
# Starting term of the arithmetic progression
a = 3

# Common difference
d = 4

# Number of terms
n = 8

print("The first 8 terms of the arithmetic progression are:")

# Loop to generate and print the terms of the arithmetic progression
for i in range(n):
    term = a + i * d
    print(term, end=" ")
```

### 10. 
Write a Python program to print the first 6 terms of a geometric sequence starting
with 2 and having a common ratio of 3.
The program should output the following sequence:
2 6 18 54 162 486

```python
# Starting term of the geometric sequence
a = 2

# Common ratio
r = 3

# Number of terms
n = 6

print("The first 6 terms of the geometric sequence are:")

# Loop to generate and print the terms of the geometric sequence
for i in range(n):
    term = a * (r ** i)
    print(term, end=" ")
```

### 11.
Write a program that asks the user for a positive integer value. The program should
calculate the sum of all the integers from 1 up to the number entered. For example, if
the user enters 20, the loop will find the sum of 1, 2, 3, 4, ... 20.

```python
# Prompt the user to enter a positive integer
n = int(input("Enter a positive integer: "))

# Initialize the sum variable
total_sum = 0

# Loop to calculate the sum of all integers from 1 to n
for i in range(1, n + 1):
    total_sum += i

# Print the result
print(f"The sum of all integers from 1 to {n} is: {total_sum}")
```

### 12.
write a program that takes a positive integer N as input and calculates the sum of
the reciprocals of all numbers from 1 up to N. The program should display the final sum.Output of the program should be like:
Enter a positive integer: 5
The sum of reciprocals from 1 to 5 is: 2.28

```python
# Import the decimal module for more precise floating-point arithmetic
from decimal import Decimal, getcontext

# Set the precision for Decimal calculations
getcontext().prec = 10

# Prompt the user to enter a positive integer
N = int(input("Enter a positive integer: "))

# Initialize the sum variable
sum_reciprocals = Decimal(0)

# Loop to calculate the sum of reciprocals from 1 to N
for i in range(1, N + 1):
    sum_reciprocals += Decimal(1) / Decimal(i)

# Print the result with 2 decimal places
print(f"The sum of reciprocals from 1 to {N} is: {sum_reciprocals:.2f}")
```

### 13.
Write a program that prompts the user to enter a number and repeats this process 5
times. The program should accumulate the numbers entered and then display the final
running total.
- Sample Output:
- Enter a number: 10
- Enter a number: 15
- Enter a number: 35
- Enter a number: 40
- Enter a number: 50
- The final running total is: 150

```python
# Initialize the total variable
running_total = 0

# Repeat the process 5 times
for _ in range(5):
    # Prompt the user to enter a number
    number = float(input("Enter a number: "))
    # Add the entered number to the running total
    running_total += number

# Print the final running total
print(f"The final running total is: {running_total}")
```

### 14.
Write a program that prompts the user to enter a positive integer and calculates its
factorial. The factorial of a positive integer 'n' is denoted as 'n!' and is calculated by
multiplying all the integers from 1 to 'n' together. For example, the factorial of 5
(denoted as 5!) is calculated as 1 x 2 x 3 x 4 x 5.
The program should display the factorial value if the input is a positive number, or
display a message stating that the factorial does not exist for negative numbers.
Additionally, for an input of zero, the program should output that the factorial of 0 is 1

```python
# Function to calculate the factorial of a positive integer
def factorial(n):
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Prompt the user to enter a positive integer
n = int(input("Enter a positive integer: "))

# Calculate and display the factorial based on the input
if n > 0:
    print(f"The factorial of {n} is: {factorial(n)}")
elif n == 0:
    print("The factorial of 0 is: 1")
else:
    print("Factorial does not exist for negative numbers.")
```

### 15.
Write a Python program that prompts the user to enter a base number and an
exponent, and then calculates the power of the base to the exponent. The program
should not use the exponentiation operator (**) or the math.pow() function. The
program should handle both positive and negative exponents.

```python
# Function to calculate the power of a base to an exponent
def power(base, exponent):
    result = 1
    if exponent >= 0:
        for _ in range(exponent):
            result *= base
    else:
        for _ in range(-exponent):
            result *= base
        result = 1 / result
    return result

# Prompt the user to enter the base number and exponent
base = float(input("Enter the base number: "))
exponent = int(input("Enter the exponent: "))

# Calculate and display the result
result = power(base, exponent)
print(f"{base} raised to the power of {exponent} is: {result}")
```
